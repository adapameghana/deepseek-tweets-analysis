```python
total_views = df['viewCount'].sum()
total_likes = df['likeCount'].sum()

# Calculate percentages
total = total_views + total_likes
percentage_views = (total_views / total) * 100
percentage_likes = (total_likes / total) * 100

# Print the percentages
print(f"Percentage of Views: {percentage_views:.2f}%")
print(f"Percentage of Likes: {percentage_likes:.2f}%")

# Plot the pie chart
plt.pie([total_views, total_likes], labels=['Views', 'Likes'], autopct='%1.1f%%')
plt.title("Total Views vs Likes")
plt.show()
```

```python
# firstly all the retweets likecount, view count,guoteCount and reply count  all together is calculated 
# here we are creating a new column 
df['total'] = df['retweetCount'] + df['replyCount'] + df['likeCount'] + df['quoteCount'] + df['viewCount']
```

```python
# now selecting the maximum count in total column 
top_trending_user = df.loc[df['total'].idxmax(), ['pseudo_author_id', 'total']]

print("Top trending user ID:", top_trending_user['pseudo_author_id'])
print("Total Count:", top_trending_user['total'])
```

```python
top_trending_users = df.nlargest(5, 'total')[['pseudo_author_id', 'total']]
topusers=[]
totalcount=[]
for index, row in top_trending_users.iterrows():
    print("Trending user ID:", row['pseudo_author_id'])
    
    print("Total Count:", row['total'])
    topusers.append(row['pseudo_author_id'])  # Storing user ID
    totalcount.append(row['total'])  # Storing total count
```

```python
sns.barplot(x=topusers, y=totalcount, palette="Blues_r")

# Labels and title
plt.xlabel("Top Users", fontsize=12)
plt.ylabel("Total Count", fontsize=12)
plt.title("Top 5 Trending Users", fontsize=14)
plt.xticks(rotation=45, fontsize=10)

plt.show()
```

```python
# each bar represents the specific and it height represents the total count 
# a thin black line is called error bar which represents the users count [black starting and ending points are the total count]
# the colours : the darker shade represents the higher value.

#barplot provides a clear and concise visualization of the top 5 trending users based on a "Total Count" metric, allowing for easy identification of influential users and comparison of their engagement levels
```

```python
sns.pairplot(df[['retweetCount', 'replyCount', 'likeCount', 'quoteCount', 'viewCount']])
plt.show()
```

A pairplot is a matrix of scatterplots that helps visualize pairwise relationships between variables in a dataset.

Diagonal: The diagonal of the matrix shows the distribution of a single variable. Typically, this is represented as a histogram or a kernel density estimate (KDE).
Off-Diagonal: The off-diagonal elements are scatterplots that show the relationship between two variables. Each scatterplot represents the values of two variables for each data point.

```python
# Heatmap to check correlations
plt.figure(figsize=(8, 6))
sns.heatmap(df[['retweetCount', 'replyCount', 'likeCount', 'quoteCount', 'viewCount']].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation between Engagement Metrics")
plt.show()
```

A heatmap is a graphical representation of data where values are represented by colors. In this case, it's used to display the correlation coefficients between different engagement metrics.

Axes: The x-axis and y-axis represent the variables you're analyzing (retweetCount, replyCount, likeCount, quoteCount, and viewCount).
Color Scale: The color scale (in this case, 'coolwarm') represents the strength and direction of the correlation.
Red: Indicates a strong positive correlation (close to +1).
Blue: Indicates a strong negative correlation (close to -1).
Lighter Colors (Whites/Light Blues/Light Oranges): Indicate weak or no correlation (close to 0).

Strong Positive Correlation: The strongest correlation is between retweetCount and likeCount (0.97). This suggests that these two metrics are highly related and often occur together.
Moderate Positive Correlations: There are moderate positive correlations between retweetCount and viewCount (0.55) and between likeCount and viewCount (0.52).
Weak Correlations: The correlations between replyCount and other metrics are weak. This suggests that replyCount might be influenced by factors different from those affecting retweets, likes, and views.
quoteCount: The absence of quoteCount in the correlation matrix suggests that you might need to include it in the calculation if you want to understand its relationship with other metrics.
